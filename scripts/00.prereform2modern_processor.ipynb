```python
from prereform2modern import Processor
from pathlib import Path

# Путь к исходному файлу
file_path = Path("ВСТАВИТЬ ПУТЬ, НАПРИМЕР:/Iskra_dataset/parvus/parvus_onp1_o_i_socrev_razvitie_orus.txt")

# Читаем текст
with file_path.open(encoding="utf-8") as f:
    orig_text = f.read()

# Преобразуем текст
text_res, changes, s_json = Processor.process_text(
    text=orig_text,
    show=False,
    delimiters=False,
    check_brackets=False
)

# Новый путь для сохранения (замена _orus на _TOCHECK)
new_file_path = file_path.parent / file_path.name.replace("_orus", "_TOCHECK")

# Сохраняем преобразованный текст
with new_file_path.open("w", encoding="utf-8") as f:
    f.write(text_res)

print(f"Файл сохранён: {new_file_path}")
```
